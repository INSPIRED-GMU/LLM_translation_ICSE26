```python
import requests
import time

# Replace with your personal access token


def github_request(url, params=None):
    """
    A helper function to perform GET requests to the GitHub API,
    returning both the JSON data and the response headers.
    If an HTTP error occurs, it prints the response text for debugging.
    """
    headers = {
       
    }
    try:
        response = requests.get(url, headers=headers, params=params)
        response.raise_for_status()
        return response.json(), response.headers
    except requests.exceptions.HTTPError as e:
        print(f"HTTP error: {e}")
        if e.response is not None:
            print(f"Response text: {e.response.text}")
        raise

def fetch_repositories(query, max_pages=10, per_page=100):
    """
    Fetch up to max_pages pages of repositories matching 'query'.
    Each page can have up to 'per_page' items (max 100).
    Returns a list of repo items (dicts).
    """
    url = "https://api.github.com/search/repositories"
    all_items = []
    
    for page in range(1, max_pages + 1):
        params = {
            "q": query,
            "sort": "stars",
            "order": "desc",
            "per_page": per_page,
            "page": page
        }
        data, headers = github_request(url, params=params)
        items = data.get("items", [])
        
        if not items:
            break
        
        all_items.extend(items)
        
        # If fewer than 'per_page', no more pages
        if len(items) < per_page:
            break
        
        # Small delay helps avoid secondary rate limiting
        time.sleep(1)
    
    return all_items

def search_issues_in_repo(owner, repo, issue_query="translate readme in:title is:issue", max_pages=3, per_page=30):
    """
    Searches for issues in a specific repo, matching issue_query 
    (default: 'translate readme in:title is:issue').
    
    Returns a list of issues (each is a dict with keys like 'title', 'html_url', etc.).
    """
    url = "https://api.github.com/search/issues"
    all_issues = []
    
    for page in range(1, max_pages + 1):
        params = {
            "q": f"repo:{owner}/{repo} {issue_query}",
            "sort": "created",
            "order": "desc",
            "per_page": per_page,
            "page": page
        }
        data, headers = github_request(url, params=params)
        
        items = data.get("items", [])
        if not items:
            break
        
        all_issues.extend(items)
        
        if len(items) < per_page:
            break
        
        time.sleep(1)
    
    return all_issues

def main():
    # 1. Get repositories with >10k stars & forks (adjust as needed)
    repo_query = "stars:>10000 forks:>10000"
    repos = fetch_repositories(repo_query, max_pages=5, per_page=100)
    
    total_repos = len(repos)
    print(f"Found {total_repos} total repos matching {repo_query}.\n")

    # 2. We'll keep a global total of all issues found
    global_issue_count = 0

    # 3. Process all repositories in batches of 10
    batch_size = 10
    for start_index in range(0, total_repos, batch_size):
        batch = repos[start_index : start_index + batch_size]
        batch_end_index = start_index + len(batch)

        print(f"Processing repos {start_index + 1} through {batch_end_index} (of {total_repos})...")

        # 4. For each repo in the current batch, search for "translate readme" issues
        for i, repo_info in enumerate(batch, start=start_index + 1):
            full_name = repo_info.get("full_name")
            if not full_name:
                continue
            
            owner, repo_name = full_name.split("/")
            print(f"{i}. Checking repo '{full_name}' for issues containing 'translate readme'...")

            issues = search_issues_in_repo(owner, repo_name, max_pages=2, per_page=30)
            issue_count = len(issues)
            global_issue_count += issue_count

            if issue_count > 0:
                print(f"   Found {issue_count} matching issue(s):")
                for issue in issues:
                    title = issue.get("title")
                    issue_url = issue.get("html_url")
                    print(f"      - {title} ({issue_url})")
            else:
                print("   No matching issues found.")
            
            print("-" * 60)
            # Optional small sleep between each repo
            time.sleep(2)

        print("Finished this batch.\n")
        # Optional: sleep after each batch of 10
        time.sleep(5)

    # 5. After all batches are processed, print the grand total
    print(f"Overall total 'translate readme' issues found across all large repos: {global_issue_count}")

if __name__ == "__main__":
    main()
```
